### 查詢帳號為A01與A02的帳號、姓名、密碼與生日資料。
2026-06-04 14:20:40

```sql
SELECT
  uid,
  cname,
  password,
  birthday
FROM
  UserInfo
WHERE
  uid IN ('A01', 'A02')
```

### 查詢姓王的基本資料
2026-06-04 14:27:25

```sql
SELECT
  uid,
  cname,
  password,
  birthday
FROM
  UserInfo
WHERE
  cname LIKE '王%'
```

## 查詢除了王大明以外的所有資料

```sql
select *
from UserInfo
where cname <>'王大明'
```
